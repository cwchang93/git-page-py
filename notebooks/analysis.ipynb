# Beginner-Friendly Iris Data Analysis

This notebook demonstrates a simple data analysis workflow using **pandas**, **numpy**, and **matplotlib**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## 1. Dataset Loading

We load the Iris dataset from an online CSV source.

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)
df.head()
```

## 2. Basic EDA

Let's inspect shape, columns, missing values, and summary statistics.

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
print('\nMissing values by column:')
print(df.isnull().sum())

df.describe()
```

## 3. Histogram Visualization

We plot histograms for all numeric columns to understand distributions.

```python
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(figsize=(10, 6), bins=15, edgecolor='black')
plt.suptitle('Iris Numeric Feature Distributions')
plt.tight_layout()
plt.show()
```

## 4. Groupby Analysis

Now we group by species and compute mean values for each numeric feature.

```python
group_summary = df.groupby('species')[numeric_cols].mean()
group_summary
```

## 5. Final Conclusion

- The dataset is clean with no missing values.
- Numeric features show different distributions across measurements.
- Grouped averages indicate clear differences among species, especially in petal length and petal width.
- This simple workflow is a strong starting point for beginner data analysis projects.

# 初學者友善的 Iris 資料分析

本 Notebook 示範如何使用 **pandas**、**numpy**、**matplotlib** 進行簡單的資料分析流程。

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## 1. 資料集載入

我們會從線上的 CSV 來源載入 Iris 資料集。

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)
df.head()
```

## 2. 基本探索式資料分析 (EDA)

接著檢查資料筆數與欄位數、欄位名稱、缺失值，以及摘要統計資訊。

```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
print('\nMissing values by column:')
print(df.isnull().sum())

df.describe()
```

## 3. 直方圖視覺化

我們為所有數值欄位繪製直方圖，以了解其分布情況。

```python
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(figsize=(10, 6), bins=15, edgecolor='black')
plt.suptitle('Iris Numeric Feature Distributions')
plt.tight_layout()
plt.show()
```

## 4. 依物種進行群組分析

接著依照物種分組，並計算各數值特徵的平均值。

```python
group_summary = df.groupby('species')[numeric_cols].mean()
group_summary
```

## 5. 結論

- 此資料集相當乾淨，沒有缺失值。
- 各數值特徵在不同測量項目上呈現不同分布。
- 群組平均值顯示不同物種之間有明顯差異，尤其在花瓣長度與花瓣寬度。
- 這個簡單流程是初學者開始資料分析專案的良好起點。
